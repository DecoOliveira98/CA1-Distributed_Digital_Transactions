### 1.1 Current System & Business Model

#### 1.1.1 The problem with current food traceability

The contemporary food supply chain is plagued by fragmentation and opacity. Food traceability currently relies on centralised systems and isolated Enterprise Resource Planning (ERP) environments, which create data silos across the network (Iqbal, 2020). In the EU, the legal baseline is the "one step back, one step forward" principle: each business only needs to identify its immediate supplier and immediate customer (European Parliament and Council of the European Union, 2002). As a result, no single participant holds an end-to-end view of a product's journey, and tracing the origin of contaminated goods during an outbreak often requires sorting through disparate paper records and incompatible digital systems.

The consequences of this fragmentation are significant:

- **Slow recalls:** before adopting blockchain, Walmart needed approximately seven days to trace the origin of a package of sliced mangoes (Kamath, 2018).
- **Public health impact:** an estimated 600 million people fall ill and 420,000 die every year from contaminated food (World Health Organization, 2015), so every hour lost in a traceback matters.
- **Food fraud:** food fraud is estimated to cost the global food industry US\$30–40 billion annually (PwC, 2016). In Ireland, the 2013 horse meat incident showed how hard it is to verify complex cross-border supply chains: the Food Safety Authority of Ireland found equine DNA in 10 out of 27 beef burger products tested (37%), none of which were labelled as containing horse meat (Food Safety Authority of Ireland, 2013).
- **Lack of trust:** consumers, regulators and even supply chain partners must trust records they cannot independently verify.

#### 1.1.2 The distributed ledger solution

To resolve these vulnerabilities, supply chains are transitioning towards blockchain technology. A blockchain is technically defined as a peer-to-peer distributed ledger that is cryptographically secure, append-only, and updateable exclusively via network consensus (Iqbal, 2020). By adopting this decentralised infrastructure, all participants (farmers, processors, logistics providers and retailers) share an identical, immutable replica of the ledger. The efficacy of this solution was notably demonstrated by Walmart and the IBM Food Trust initiative: by using a shared distributed ledger, the mango traceback time was reduced from about seven days to 2.2 seconds (Kamath, 2018). This immediate visibility allows targeted product recalls, safeguarding public health without the blanket destruction of unaffected inventory. Reviews of the field confirm a rapid growth of blockchain initiatives across agriculture and food supply chains, while noting that most are still at an early stage of maturity (Kamilaris, Fonts and Prenafeta-Boldú, 2019).

#### 1.1.3 Cryptographic security

The integrity of this traceability network depends on cryptographic primitives. While hashing (e.g. in Merkle trees) links data blocks together securely, asymmetric key cryptography is the core concept that gives identity to participants and proves ownership of assets (Iqbal, 2020). Specifically, Elliptic-Curve Cryptography (ECC) provides strong security with small key sizes. Through curves such as secp256k1, each supply chain participant generates a mathematically related private and public key pair (Shrivastava, Le and Sharma, 2020). The private key allows a supplier to digitally sign a transaction (e.g. confirming a crop's harvest date), while the public key allows the rest of the network to authenticate that claim trustlessly. *(Expanded in section 1.4.)*

#### 1.1.4 Proposed business model

The proposed solution is a **consortium (permissioned) blockchain**, governed jointly by the supply chain members rather than by a single company. Only registered participants can write data, but consumers and regulators can read the history of any batch.

| Actor | Role on the blockchain | Contribution / cost | Value gained |
|---|---|---|---|
| Farmer | Creates the batch (origin, harvest date, certifications) | Low-cost access through a DApp / mobile app | Proof of origin and quality, potential premium pricing, protection from unfair blame in a recall |
| Processor | Records processing steps and links input batches to products | Subscription / transaction fees | Automated compliance records, lower audit costs |
| Distributor | Transfers batches and records checkpoints (location, temperature via IoT) | Subscription; may host a node | Proof of correct handling, fewer disputes |
| Retailer | Receives the final product; main sponsor and node operator | Funds most of the infrastructure | Targeted recalls, less waste, stronger brand trust |
| Regulator / auditor | Read access to the full history | None | Real-time, verifiable audits |
| Consumer | Scans a QR code on the package | Free | Verifies origin and journey of the product |

**Revenue model:** the platform is offered as Blockchain-as-a-Service with subscription fees scaled to company size. IBM Food Trust followed this model at launch, charging between US\$110 and US\$11,000 per month depending on the size of the business (Canellis, 2018). Large retailers and distributors therefore finance the core infrastructure, while smaller producers join at low cost. The return on investment comes from smaller and faster recalls, reduced fraud, lower audit costs and the ability to sell verified products at a premium.

#### 1.1.5 Disruptive influence on existing systems

This technological shift fundamentally disrupts the traditional business model of supply chains:

- **From data silos to a single source of truth:** instead of each company keeping its own records, all participants write to one shared, tamper-evident ledger.
- **Disintermediation:** trust no longer depends on paper certificates and third-party verification alone; it is enforced by cryptography and consensus (Iqbal, 2020).
- **From reactive to proactive food safety:** recalls move from blanket withdrawals to targeted, batch-level actions within seconds (Kamath, 2018).
- **Automation through smart contracts:** rules such as "only the current owner can transfer a batch" or "flag a batch if the cold chain is broken" are executed automatically.
- **Power shift towards producers and consumers:** small producers can prove the quality of their products directly, and consumers gain visibility that was previously restricted to large companies.

However, blockchain only guarantees the integrity of data *after* it is recorded. Incorrect or fraudulent data entered at the source remains a key limitation (Galvez, Mejuto and Simal-Gandara, 2018). This is discussed further in section 1.3.

### 1.2 Proposed Blockchain Solution

#### 1.2.1 Overview

The proposed solution is a **Food Traceability DApp** built on an Ethereum-compatible blockchain, in which every product batch is registered on-chain and every change of custody is recorded as a transfer between two parties: a **sender** (the current owner) and a **receiver** (the next actor in the supply chain). The blockchain acts as the **middle layer** between users: no participant owns the data, but every participant can verify it.

The journey of a batch follows these steps:

1. The consortium administrator registers each participant's blockchain address and role (Farmer, Processor, Distributor or Retailer).
2. The **Farmer** creates a batch (product, origin, quantity) and stores the Keccak-256 hash of its quality certificate on-chain, while the certificate itself is kept off-chain.
3. The batch is transferred **Farmer → Processor → Distributor → Retailer**. Each transfer is signed by the sender and only succeeds if the sender is the current owner and the receiver is a registered participant.
4. During transport, the Distributor records **checkpoints** (location and temperature). If the temperature exceeds the safe limit, the contract emits an alert event.
5. The **Retailer** marks the batch as sold. Consumers and regulators can read the full history of the batch at any time, for example by scanning a QR code on the package.
6. If a problem is found, the administrator can **recall** the batch, which immediately blocks any further transfers and identifies every party that handled it.

#### 1.2.2 Recent advancements incorporated

- **IoT sensors for the cold chain:** temperature and GPS sensors can record conditions automatically during transport, reducing manual data entry, which is one of the main sources of error in traceability systems (Kamilaris, Fonts and Prenafeta-Boldú, 2019).
- **Blockchain oracles:** smart contracts cannot access data outside the blockchain by themselves. Oracles bring external data (e.g. sensor readings) on-chain, but they reintroduce a trusted party; this is known as the *oracle problem* (Caldarelli, 2020).
- **Off-chain storage with IPFS:** storing large files on-chain is expensive, so certificates and documents are stored on the InterPlanetary File System, a content-addressed peer-to-peer file system (Benet, 2014). Only the document's hash is stored in the smart contract, so any later change to the document can be detected.
- **Layer 2 scaling:** rollups execute transactions off the main Ethereum chain and post the results back to it, reducing fees and increasing throughput while inheriting Ethereum's security (Ethereum Foundation, 2024). This makes frequent, low-value supply chain transactions economically viable.
- **New regulation driving digital traceability:** in the US, the FDA's Food Traceability Rule (FSMA 204) requires businesses handling high-risk foods to keep Key Data Elements for Critical Tracking Events and to provide records to the FDA within 24 hours of a request, with compliance now expected by 20 July 2028 (U.S. Food and Drug Administration, n.d.). In the EU, the Deforestation Regulation requires traceability information, including the geolocation of the land of production, for commodities such as cattle, cocoa, coffee and soya (European Parliament and Council of the European Union, 2023), applying to large operators from 30 December 2026 (CIF, 2026).

### 1.3 Benefits & Challenges

| Benefits | Challenges |
|---|---|
| **Immutability:** once recorded, transactions cannot be altered or deleted without detection (Iqbal, 2020). | **"Garbage in, garbage out":** blockchain guarantees the integrity of data after it is recorded, not its accuracy at the source (Galvez, Mejuto and Simal-Gandara, 2018). |
| **Transparency & single source of truth:** all authorised participants see the same data in real time. | **Oracle problem:** external data (sensors, manual inputs) must be trusted before it reaches the chain (Caldarelli, 2020). |
| **Faster, targeted recalls:** traceback reduced from days to seconds (Kamath, 2018). | **Cost & scalability:** every write costs gas, and public blockchains have limited throughput; Layer 2 solutions mitigate this (Ethereum Foundation, 2024). |
| **Fraud prevention:** forged origin claims are harder because every record is signed by a known participant. | **Privacy between competitors:** commercially sensitive data (prices, volumes, suppliers) should not be visible to everyone, which favours permissioned blockchains and off-chain storage. |
| **Auditability & compliance:** regulators get a complete, verifiable history (e.g. supporting FSMA 204 and EUDR requirements). | **Adoption barriers:** Saberi *et al.* (2019) group barriers into four categories: intra-organisational, inter-organisational, technical and external. |
| **Consumer trust:** consumers can verify the origin of a product themselves. | **Digital divide:** small farmers may lack the devices, connectivity or skills needed to take part (Kamilaris, Fonts and Prenafeta-Boldú, 2019). |

### 1.4 Cryptography

#### 1.4.1 Chosen method: Elliptic Curve Digital Signature Algorithm (ECDSA) on secp256k1, combined with Keccak-256 hashing

The most suitable cryptographic approach for this scenario combines two primitives, both native to Ethereum (Wood, 2014):

**1. ECDSA (secp256k1) for identity and authenticity**

- Each participant has a **private key** (a random 256-bit number) and a **public key** derived from it by elliptic curve point multiplication. Reversing this operation (the elliptic curve discrete logarithm problem) is computationally infeasible (Johnson, Menezes and Vanstone, 2001).
- The participant's **blockchain address** is the last 20 bytes of the Keccak-256 hash of the public key (Wood, 2014).
- Every transaction (e.g. `transferBatch`) is **signed with the sender's private key**. The network recovers the sender's address from the signature, and the smart contract reads it as `msg.sender`.
- This is what makes the contract's access control work: only the Farmer's address can create a batch, and only the current owner can transfer it. Nobody can impersonate another participant without their private key.

**2. Keccak-256 for integrity**

- Keccak-256 is the hash function used throughout Ethereum, for transaction and block hashes and for the Merkle Patricia tries that store the blockchain state (Wood, 2014). *(Note: it differs slightly from the standardised NIST SHA3-256 in its padding.)*
- Each block contains the hash of the previous block, so changing any past record would change every following hash, making tampering immediately evident (Nakamoto, 2008).
- **Merkle trees** summarise all transactions of a block into a single root hash, allowing efficient verification that a transaction belongs to a block (Nakamoto, 2008).
- In the smart contract, the **certificate hash** links an off-chain document (stored on IPFS) to the batch: anyone can hash the document and compare it with the on-chain value using `verifyCertificate()`.

#### 1.4.2 How it enhances security

| Security property | Provided by | In the food supply chain |
|---|---|---|
| **Authenticity** | ECDSA signature | Proves *which* farmer, processor or distributor recorded each event |
| **Non-repudiation** | ECDSA signature | A participant cannot later deny having transferred or recorded a batch, which is important when assigning responsibility in a recall |
| **Integrity** | Keccak-256 hashes, chained blocks, Merkle trees | Records and certificates cannot be modified without detection |
| **Access control** | Address derived from public key (`msg.sender`) | Only authorised roles can perform each action in the smart contract |

#### 1.4.3 Why ECDSA rather than RSA?

- **Smaller keys for the same security:** a 256-bit elliptic curve key provides roughly the same security strength (128 bits) as a 3072-bit RSA key (Barker, 2020). This means smaller signatures, less storage and lower transaction costs.
- **Efficiency:** faster key generation and signing make ECC more suitable for resource-constrained devices such as IoT sensors in the cold chain.
- **Native support:** Ethereum and Bitcoin both use secp256k1, so no additional infrastructure is needed (Wood, 2014).

**Limitation:** ECDSA and hashing provide integrity and authenticity, **not confidentiality**. All data on a public blockchain is readable, which is why commercially sensitive information should stay off-chain or on a permissioned network. In the long term, large-scale quantum computers could threaten elliptic curve cryptography, which is driving research into post-quantum signature schemes.
