### 1.1 Current system & business model
### 1.1 Research 

The contemporary food supply chain is plagued by fragmentation and opacity. Currently, food traceability relies heavily on centralized systems and isolated Enterprise Resource Planning (ERP) environments, which create data silos across the network (Iqbal, 2020). Because there is no single source of truth, tracing the origin of contaminated goods during an outbreak often requires sorting through disparate paper records and incompatible digital systems. This inefficiency leads to slow recalls, widespread food fraud, and a fundamental lack of trust among consumers and stakeholders.

##### The Distributed Ledger Solution
    To resolve these vulnerabilities, supply chains are transitioning towards blockchain technology. A blockchain is technically defined as a peer-to-peer distributed ledger that is cryptographically secure, append-only, and updateable exclusively via network consensus (Iqbal, 2020). By adopting this decentralized infrastructure, all participants—farmers, logistics providers, processors, and retailers—share an exact, immutable replica of the ledger. The efficacy of this solution was notably demonstrated by Walmart and the IBM Food Trust initiative. Prior to blockchain integration, tracing the origin of sliced mangoes took Walmart approximately seven days. By utilizing a shared distributed ledger, this traceback time was reduced to a mere 2.2 seconds (Kamath, 2018). This immediate visibility allows for targeted product recalls, safeguarding public health without necessitating the blanket destruction of unaffected inventory.

##### Cryptographic Security and Business Disruption
    The integrity of this traceability network is heavily dependent on cryptographic primitives. While hashing (such as Merkle trees) links data blocks together securely, asymmetric key cryptography is the core concept that gives identity to participants and proves the ownership of assets (Iqbal, 2020). Specifically, Elliptic-Curve Cryptography (ECC) provides optimal security for these networks. Through curves such as secp256k1, each supply chain participant generates a mathematically related private and public key pair (Shrivastava, Le and Sharma, 2020). The private key allows a supplier to digitally sign a transaction (e.g., verifying a crop's harvest date), while the public key allows the rest of the network to authenticate that claim trustlessly.

    This technological shift fundamentally disrupts the traditional business model of supply chains. In this proposed framework, large retailers and distributors typically finance the core infrastructure and cloud-based node hosting, while smaller participants access the network via decentralized applications (DApps). By replacing trusted third parties with a cryptographic, consensus-driven environment (Iqbal, 2020), the industry shifts from reactive auditing to proactive, verifiable food safety.
